```python
name = 'REDROSE'
choice =int(input('Enter the Number:'))

if choice == 1:
  for  i in range(1, len(name) + 1 ):
    print(name[:i])

elif choice == 2:
  for i in range(1, len(name) + 1):
    print(" " * (len(name) - i) +" ".join(name[:i]))

elif choice == 3:
  for i in range (1,len(name) + 1 ):
    print(" " * (len(name) -i) + name[:i])
```
